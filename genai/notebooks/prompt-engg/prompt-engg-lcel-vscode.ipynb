### 1. What is LCEL & Why Use It?

LangChain Expression Language (LCEL) uses the pipe operator (`|`) to link prompts, models, and parsers sequentially.

### 2. LCEL Code Examples

#### Example 1: Basic QA Chain

````python
from langchain_core.prompts import PromptTemplate
from langchain.chat_models import init_chat_model

# Initialize components
llm = init_chat_model("gemini-2.5-flash", model_provider="google_genai")
prompt = PromptTemplate.from_template("Tell me a short joke about {topic}.")

# Compose LCEL Chain
chain = prompt | llm

# Invoke the chain
response = chain.invoke({"topic": "programming"})
print(response.content)
````

#### Example 2: Subject Line Generator

````python
prompt2 = PromptTemplate.from_template(
    "Write a single email subject line for a request about {request_topic} in a {tone} tone."
)
chain2 = prompt2 | llm

response = chain2.invoke({"request_topic": "server maintenance delay", "tone": "apologetic"})
print(response.content)
````

### 4. Invoking vs. Streaming

#### 4.1 invoke() Example

````python
response = chain.invoke({"topic": "Python"})
print(response.content)
````

#### 4.2 stream() Example

````python
for chunk in chain.stream({"topic": "Python generators"}):
    print(chunk.content, end="", flush=True)
````

### 5. LangChain Runnables

* **`RunnablePassthrough`**: Forwards values as-is.
* **`RunnableParallel`**: Executes runnables concurrently.

#### 5.2 RunnablePassthrough Example

````python
from langchain_core.runnables import RunnablePassthrough, RunnableLambda

chain_pass = {
    "original": RunnablePassthrough(),
    "upper": RunnableLambda(lambda x: x.upper())
}
print(chain_pass.invoke("hello"))
````

#### 5.3 RunnableParallel Example

````python
from langchain_core.runnables import RunnableParallel

prompt1 = PromptTemplate.from_template("Summarize {topic}")
prompt2 = PromptTemplate.from_template("List advantages of {topic}")

parallel_chain = RunnableParallel(
    summary=prompt1 | llm,
    advantages=prompt2 | llm
)

result = parallel_chain.invoke({"topic": "Artificial Intelligence"})
print("Summary:", result['summary'].content)
print("Advantages:", result['advantages'].content)
````
