### Pydantic Output Parser - Practice & Exercises

In this notebook, you will practice defining schemas using Pydantic's `BaseModel` and extracting strongly-typed Python objects from LLM responses.

#### Part 1: Follow-Along Practice

##### Setting up the Pydantic Schema and Parser

````python
from dotenv import load_dotenv
from langchain.prompts import ChatPromptTemplate
from langchain.output_parsers import PydanticOutputParser
from langchain.chat_models import init_chat_model
from pydantic import BaseModel, Field

# Load environment variables
load_dotenv()

# Initialize Groq model
model = init_chat_model("llama-3.3-70b-versatile", model_provider="groq")

# Define the schema
class PersonProfile(BaseModel):
    name: str = Field(description="The full name of the person")
    occupation: str = Field(description="The primary job/occupation of the person")
    skills: list[str] = Field(description="A list of 3-5 core professional skills")
    years_of_experience: int = Field(description="Estimated years of professional experience in their field")

# Instantiate the parser
pydantic_parser = PydanticOutputParser(pydantic_object=PersonProfile)
format_instructions = pydantic_parser.get_format_instructions()

# Construct prompt
prompt = ChatPromptTemplate.from_messages([
    ("system", "Generate information based on the input name. {format_instructions}"),
    ("human", "Generate a profile for 'Ada Lovelace'.")
])

# Format, Invoke, and Parse
messages = prompt.format_messages(format_instructions=format_instructions)
response = model.invoke(messages)
result = pydantic_parser.parse(response.content)

print(f"Name: {result.name}")
print(f"Occupation: {result.occupation}")
print(f"Skills: {result.skills}")
print(f"Experience: {result.years_of_experience}")
````

from pydantic import BaseModel, Field
from langchain.output_parsers import PydanticOutputParser
from langchain.prompts import ChatPromptTemplate
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv

load_dotenv()
model = init_chat_model("llama-3.3-70b-versatile", model_provider="groq")

class Movie(BaseModel):
    title: str = Field(description="Title of the movie")
    director: str = Field(description="Director of the movie")
    release_year: int = Field(description="Year the movie was released")

parser = PydanticOutputParser(pydantic_object=Movie)
instructions = parser.get_format_instructions()

prompt = ChatPromptTemplate.from_messages([
    ("system", "Provide movie details. {format_instructions}"),
    ("human", "Details for the movie Interstellar.")
])

messages = prompt.format_messages(format_instructions=instructions)
response = model.invoke(messages)
movie = parser.parse(response.content)

print(movie)
print(f"Title: {movie.title}, Director: {movie.director}")
