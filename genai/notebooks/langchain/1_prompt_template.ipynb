### PromptTemplate - Practice & Exercises

In this notebook, you will practice formatting plain string prompt templates using LangChain's `PromptTemplate` and invoking them using core abstractions (Groq provider).

#### Part 1: Follow-Along Practice

##### Method 1: Using `.from_template()`
Define a template string with placeholders and construct it using `.from_template()`.

````python
# Load environment variables from .env file
from dotenv import load_dotenv
load_dotenv()

# Import PromptTemplate from langchain
from langchain.prompts import PromptTemplate
# Import core abstraction to initialize chat model
from langchain.chat_models import init_chat_model

# Initialize Groq model using core abstractions
model = init_chat_model("llama-3.3-70b-versatile", model_provider="groq")

# Define a simple template with a topic variable
template_str = "Tell me a joke about {topic}."
# Instantiate via from_template
prompt_template = PromptTemplate.from_template(template_str)

# Format and invoke the prompt template
prompt_value = prompt_template.invoke({"topic": "cats"})
# Run the model with the formatted prompt value
response = model.invoke(prompt_value)
print(response.content)
````

##### Method 2: Direct Class Constructor
Initialize `PromptTemplate` by specifying inputs and the template explicitly.

````python
# Import PromptTemplate from langchain
from langchain.prompts import PromptTemplate

# Initialize PromptTemplate directly using constructor variables
prompt_template_direct = PromptTemplate(
    input_variables=["adjective", "animal"],
    template="Tell me a {adjective} story about a {animal}."
)

# Format and invoke the prompt template directly
prompt_value = prompt_template_direct.invoke({"adjective": "funny", "animal": "panda"})
# Run the model with the formatted prompt value
response = model.invoke(prompt_value)
print(response.content)
````

from langchain.prompts import PromptTemplate
from langchain.chat_models import init_chat_model
