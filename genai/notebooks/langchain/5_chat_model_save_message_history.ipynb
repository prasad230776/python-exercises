## Chat Model Save Message History (SQLite) - Practice & Exercises
In this notebook, you will learn how to persist conversation histories inside a local SQLite database using LangChain's SQLChatMessageHistory, and retrieve them later.

### Follow-Along Practice: Persist Chat Logs
#### GoalInitialize an SQL-backed chat message history and append conversational turns.

#### Plan1. Import `SQLChatMessageHistory` from `langchain_community.chat_message_histories`.
2. Instantiate `SQLChatMessageHistory` with a `session_id` and a local database path.
3. Append human and AI messages and verify they are stored.

````python
from dotenv import load_dotenv
from langchain_community.chat_message_histories import SQLChatMessageHistory
from langchain.chat_models import init_chat_model

load_dotenv()

# Setup session
SESSION_ID = "user_session_sqlite"
DB_CONNECTION = "sqlite:///chat_history.db"

chat_history = SQLChatMessageHistory(
    session_id=SESSION_ID,
    connection_string=DB_CONNECTION
)

model = init_chat_model("llama-3.3-70b-versatile", model_provider="groq")

# Simulate adding a round of conversation
chat_history.add_user_message("Hello! My favorite color is Red.")
response = model.invoke(chat_history.messages)
chat_history.add_ai_message(response.content)

print("AI Response:", response.content)
print("Saved History:", chat_history.messages)
````
