### Graph 0: Simple Hello World (No Name Input)

#### Goal
We want to build a simple graph that updates a state containing only one variable `msg` to `"Hello, World!"` without requiring custom name input.

#### Steps:
1. Define the `AgentState` schema with a single variable `msg`.
2. Define the node function `hello_node` that sets `msg` to `"Hello, World!"`.
3. Build and compile the graph.
4. Invoke the graph with an initial empty message.

````python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END

# 1. Define the State Schema
class AgentState0(TypedDict):
    msg: str
````

````python
# 2. Define the Node Function
def hello_node(state: AgentState0) -> AgentState0:
    state['msg'] = "Hello, World!"
    return state
````

````python
# 3. Build and Compile the Graph
graph0 = StateGraph(AgentState0)
graph0.add_node("hello_world", hello_node)
graph0.set_entry_point("hello_world")
graph0.set_finish_point("hello_world")

app0 = graph0.compile()
````

````python
# 4. Invoke the Graph
result0 = app0.invoke({"msg": ""})
print(result0)
````

### Graph I: Hello World Graph (With Name Input)

#### Goal
We want to build a simple graph that takes a user's name as input (e.g., `{"name": "Bob"}`) and outputs a personalized greeting (e.g., `{"greet": "Hey Bob, how is your day going?"}`).

#### Steps:
1. Define the `AgentState` schema containing `name` and `greet` variables.
2. Define the node function `greeting_node` that reads the input `name` and constructs the personalized greeting.
3. Build and compile the graph.
4. Invoke the graph with a custom name.

````python
# 1. Define the State Schema
class AgentState(TypedDict):
    name: str
    greet: str
````

````python
# 2. Define the Node Function
def greeting_node(state: AgentState) -> AgentState:
    state['greet'] = f"Hey {state['name']}, how is your day going?"
    return state
````

````python
# 3. Build and Compile the Graph
graph = StateGraph(AgentState)
graph.add_node("greeter", greeting_node)
graph.set_entry_point("greeter")
graph.set_finish_point("greeter")

app = graph.compile()
````

````python
# 4. Invoke the Graph
result = app.invoke({"name": "Bob"})
print(result)
````
