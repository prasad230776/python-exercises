#### Part 1: Follow-Along Practice
##### GoalBuild a simple stateless agent using LangGraph that takes user messages and passes them directly to an LLM for processing.

##### Steps:1. Define the `AgentState` schema containing `messages` (a list of messages).
2. Define the `process` node function that invokes the LLM with the messages.
3. Build the graph by adding the node and connecting START to it and it to END.
4. Compile and invoke the agent.

````python
from typing import TypedDict
from langchain_core.messages import HumanMessage
from langchain_openai import ChatOpenAI
from langgraph.graph import StateGraph, START, END
from dotenv import load_dotenv

load_dotenv()

# 1. Define the State Schema
class AgentState(TypedDict):
    messages: list[HumanMessage]
````

````python
# 2. Define the Node Function
llm = ChatOpenAI(model="gpt-4o")

def process(state: AgentState) -> AgentState:
    response = llm.invoke(state["messages"])
    print(f"\nAI: {response.content}")
    return state
````

````python
# 3. Build and Compile the Graph
graph = StateGraph(AgentState)
graph.add_node("process", process)
graph.add_edge(START, "process")
graph.add_edge("process", END) 
agent = graph.compile()
````

````python
# 4. Invoke the Agent
user_msg = [HumanMessage(content="Tell me a quick 1-line joke.")]
agent.invoke({"messages": user_msg})
````
