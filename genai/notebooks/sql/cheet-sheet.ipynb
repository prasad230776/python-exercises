#### Database Setup & Initialization

##### 1. Database Table Creation (DDL)
Execute the code cell below to connect to `ecommerce.db` and create the master relational schema tables (`customers`, `products`, `orders`, `order_items`).

<details>
<summary>Click to view Table Creation SQL (DDL)</summary>

```sql
CREATE TABLE customers (
    customer_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    city TEXT NOT NULL,
    state TEXT NOT NULL,
    join_date TEXT NOT NULL
);

CREATE TABLE products (
    product_id INTEGER PRIMARY KEY,
    product_name TEXT NOT NULL,
    category TEXT NOT NULL,
    unit_price REAL NOT NULL
);

CREATE TABLE orders (
    order_id INTEGER PRIMARY KEY,
    customer_id INTEGER NOT NULL,
    order_date TEXT NOT NULL,
    status TEXT NOT NULL,
    FOREIGN KEY (customer_id) REFERENCES customers(customer_id)
);

CREATE TABLE order_items (
    item_id INTEGER PRIMARY KEY,
    order_id INTEGER NOT NULL,
    product_id INTEGER NOT NULL,
    quantity INTEGER NOT NULL,
    item_price REAL NOT NULL,
    FOREIGN KEY (order_id) REFERENCES orders(order_id),
    FOREIGN KEY (product_id) REFERENCES products(product_id)
);
```
</details>


````python
import sqlite3
import pandas as pd

# Connect to database
conn = sqlite3.connect("ecommerce.db")
cursor = conn.cursor()

# Enable foreign keys
cursor.execute("PRAGMA foreign_keys = ON;")

# Create relational tables
cursor.executescript('''
DROP TABLE IF EXISTS order_items;
DROP TABLE IF EXISTS orders;
DROP TABLE IF EXISTS products;
DROP TABLE IF EXISTS customers;

CREATE TABLE customers (
    customer_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    city TEXT NOT NULL,
    state TEXT NOT NULL,
    join_date TEXT NOT NULL
);

CREATE TABLE products (
    product_id INTEGER PRIMARY KEY,
    product_name TEXT NOT NULL,
    category TEXT NOT NULL,
    unit_price REAL NOT NULL
);

CREATE TABLE orders (
    order_id INTEGER PRIMARY KEY,
    customer_id INTEGER NOT NULL,
    order_date TEXT NOT NULL,
    status TEXT NOT NULL,
    FOREIGN KEY (customer_id) REFERENCES customers(customer_id)
);

CREATE TABLE order_items (
    item_id INTEGER PRIMARY KEY,
    order_id INTEGER NOT NULL,
    product_id INTEGER NOT NULL,
    quantity INTEGER NOT NULL,
    item_price REAL NOT NULL,
    FOREIGN KEY (order_id) REFERENCES orders(order_id),
    FOREIGN KEY (product_id) REFERENCES products(product_id)
);
''')

conn.commit()
print("Master tables created successfully in ecommerce.db!")
````

#### Chapter 2: Data Definition Language - DDL

##### Schema Creation & Optimization Tasks


**Exercise 2.1: Table Creation with Constraints**

**Task**: Define a new table named product_reviews with primary key, foreign key referencing products, non-null rating check constraint (1 to 5), and default review dates.

**Execution Plan**:
* Use `CREATE TABLE IF NOT EXISTS product_reviews (...)`.
* Add `CHECK (rating BETWEEN 1 AND 5)` constraint.

<details>
<summary>Click to view Solution SQL</summary>

```sql
CREATE TABLE IF NOT EXISTS product_reviews (
    review_id INTEGER PRIMARY KEY AUTOINCREMENT,
    product_id INTEGER NOT NULL,
    rating INTEGER CHECK (rating BETWEEN 1 AND 5),
    review_text TEXT,
    review_date TEXT DEFAULT (CURRENT_DATE),
    FOREIGN KEY (product_id) REFERENCES products(product_id)
);
```
</details>

**Execution Plan**:

* Use `CREATE UNIQUE INDEX` for composite uniqueness.

<details>
<summary>Click to view Solution SQL</summary>

```sql
CREATE INDEX idx_orders_customer_id ON orders(customer_id);
```

</details>


**Exercise 2.3: Create Virtual Views**

**Task**: Build a reusable virtual view v_completed_orders joining customer names and cities with order headers.

**Execution Plan**:
* Use `CREATE VIEW v_completed_orders AS SELECT ...` joining `orders` with `customers`.

<details>
<summary>Click to view Solution SQL</summary>

```sql
CREATE VIEW v_completed_orders AS
SELECT 
    o.order_id,
    c.name AS customer_name,
    c.city,
    o.order_date
FROM orders o
JOIN customers c ON o.customer_id = c.customer_id
WHERE o.status = 'Completed';
```
</details>

```sql
    SELECT o.order_date, SUM(oi.quantity * oi.item_price) AS daily_revenue
    FROM orders o
    JOIN order_items oi ON o.order_id = oi.order_id
    WHERE o.status = 'Completed'
    GROUP BY o.order_date
)
SELECT order_date, daily_revenue, SUM(daily_revenue) OVER (ORDER BY order_date ASC) AS running_total_revenue FROM DailyRevenue ORDER BY order_date ASC;
```

**Query 27: Repeat Purchase Recency Gap (LAG() OVER)**

**Task**: Calculate the gap in days between a customer's current order date and their previous order date using LAG().

**Execution Plan**:
* Select `customer_id`, `order_id`, and `order_date` from `orders`.
* Apply `LAG(order_date) OVER (PARTITION BY customer_id ORDER BY order_date ASC)`.
* Compute `JULIANDAY(order_date) - JULIANDAY(previous_order_date)`.

<details>
<summary>Click to view Solution SQL</summary>

```sql
WITH OrderHistory AS (
    SELECT order_id, customer_id, order_date, LAG(order_date) OVER (PARTITION BY customer_id ORDER BY order_date ASC) AS previous_order_date
    FROM orders
)
SELECT order_id, customer_id, order_date, previous_order_date, CAST((JULIANDAY(order_date) - JULIANDAY(previous_order_date)) AS INTEGER) AS days_since_last_order FROM OrderHistory WHERE previous_order_date IS NOT NULL;
```
