#### Exercise 1: Simulated Async File Downloader

**Task:**
1. Define an asynchronous coroutine `download_file(file_id, delay)` that prints `"Downloading file {file_id}..."`, waits for `delay` seconds using `asyncio.sleep()`, and then prints `"Finished downloading file {file_id}"` and returns `f"file_{file_id}_data"`.
2. Write an async function `main()` that runs three file downloads concurrently using `asyncio.gather()`:
   - File 1: delay of 3 seconds
   - File 2: delay of 1 second
   - File 3: delay of 2 seconds
3. Measure the execution time of the entire process to confirm they run concurrently.

```python
import asyncio
import time

async def download_file(file_id, delay):
    print(f"Downloading file {file_id}...")
    await asyncio.sleep(delay)
    print(f"Finished downloading file {file_id}")
    return f"file_{file_id}_data"

async def main():
    start_time = time.time()
    
    results = await asyncio.gather(
        download_file(1, 3),
        download_file(2, 1),
        download_file(3, 2)
    )
    
    end_time = time.time()
    print(f"Results: {results}")
    print(f"Total elapsed time: {end_time - start_time:.2f} seconds")

await main()  # Use asyncio.run(main()) in standard scripts
```

#### Exercise 2: Concurrent Batch File Manager

**Task:**
1. Define a function `process_file(filename)` that simulates a blocking/synchronous read operation (e.g. using `time.sleep(2)`).
2. Implement a multithreaded run using `threading.Thread` to process a list of filenames `["file_a.txt", "file_b.txt", "file_c.txt"]` concurrently.
3. Ensure the main program waits for all threads to complete using `join()` before printing `"All files processed"`.

```python
import threading
import time

def process_file(filename):
    print(f"Processing {filename}...")
    time.sleep(2)  # Simulates blocking I/O
    print(f"Finished {filename}")

filenames = ["file_a.txt", "file_b.txt", "file_c.txt"]
threads = []

# Create and start threads
for name in filenames:
    t = threading.Thread(target=process_file, args=(name,))
    threads.append(t)
    t.start()

# Wait for all threads to finish
for t in threads:
    t.join()

print("All files processed")
```
