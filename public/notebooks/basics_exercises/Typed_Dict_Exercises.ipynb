### Exercise 3: Union Types with Optional Values
**Task:** Define a `User` `TypedDict` where `username` (str) and `email` (str | None) are required. Create an instance with `email` as `None` and another with a string email.

````python
# TODO: Define the User TypedDict and create valid instances below
````

<details>
<summary>💡 Click to see solution</summary>

```python
from typing import TypedDict

class User(TypedDict):
    username: str
    email: str | None

# Valid: email is None
user1: User = {
    "username": "alice",
    "email": None
}

# Valid: email is a string
user2: User = {
    "username": "bob",
    "email": "bob@example.com"
}
print(user1)
print(user2)
```

### Exercise 4: NotRequired Keys in a Book Schema
**Task:** Define a `Book` `TypedDict` where `title` (str) and `author` (str) are required, but `pages` (int) and `genre` (str | None) are `NotRequired`. Create an instance without `pages` and `genre`, and another with both.

````python
# TODO: Define the Book TypedDict and create instances below
````

<details>
<summary>💡 Click to see solution</summary>

```python
from typing import TypedDict, NotRequired

class Book(TypedDict):
    title: str
    author: str
    pages: NotRequired[int]
    genre: NotRequired[str | None]

# Valid: Pages and genre omitted
book1: Book = {
    "title": "1984",
    "author": "George Orwell"
}

# Valid: Pages and genre provided
book2: Book = {
    "title": "To Kill a Mockingbird",
    "author": "Harper Lee",
    "pages": 281,
    "genre": None
}
print(book1)
print(book2)
```
</details>

### Exercise 5: LLMState Schema for LangGraph
**Task:** Define an `LLMState` `TypedDict` for a translation agent with required key `original_text` (str), and optional keys `source_language` (NotRequired[str]), `target_language` (NotRequired[str]), and `translated_text` (NotRequired[str]). Create instances representing the state at different stages.

````python
# TODO: Define the LLMState TypedDict and create sequential states below
````

<details>
<summary>💡 Click to see solution</summary>

```python
from typing import TypedDict, NotRequired

class LLMState(TypedDict):
    original_text: str
    source_language: NotRequired[str]
    target_language: NotRequired[str]
    translated_text: NotRequired[str]

# Stage 1: Initial state
state1: LLMState = {
    "original_text": "Bonjour tout le monde"
}

# Stage 2: Languages detected/defined
state2: LLMState = {
    "original_text": "Bonjour tout le monde",
    "source_language": "French",
    "target_language": "English"
}

# Stage 3: Translated text populated
state3: LLMState = {
    "original_text": "Bonjour tout le monde",
    "source_language": "French",
    "target_language": "English",
    "translated_text": "Hello everyone"
}
print("Stage 1:", state1)
print("Stage 2:", state2)
print("Stage 3:", state3)
```
