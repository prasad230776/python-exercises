# TypedDict - Follow-Along Practice
This notebook covers defining expected dictionary structures, required keys, union types, optional keys with `NotRequired`, and modeling LangGraph state.

### 1. What is TypedDict?
Define the expected structure of a dictionary using type hints. It is mainly for static type checking and doesn't enforce runtime checks.

```python
from typing import TypedDict

class Student(TypedDict):
    name: str
    age: int
    marks: float

# Create a student dictionary
student: Student = {
    "name": "Ravi",
    "age": 21,
    "marks": 85.5
}
print(student)
```

### 2. Required Keys
By default, all keys defined in a `TypedDict` are required.

```python
class StudentRequired(TypedDict):
    name: str
    age: int

student_info: StudentRequired = {
    "name": "Ravi",
    "age": 21
}
print(student_info)
```

### 3. Union Types with `|`
Use the pipe `|` operator to allow a key to accept multiple value types (like `int | None`), while keeping the key itself required.

```python
class StudentUnion(TypedDict):
    name: str
    age: int | None

student_a: StudentUnion = {"name": "Ravi", "age": 21}
student_b: StudentUnion = {"name": "Ravi", "age": None}
print(student_a)
print(student_b)
```

### 4. Optional Keys with `NotRequired`
Use `NotRequired` to indicate that a key itself can be completely omitted from the dictionary.

```python
from typing import NotRequired

class StudentOptional(TypedDict):
    name: str
    age: NotRequired[int]

student_with_age: StudentOptional = {"name": "Ravi", "age": 21}
student_no_age: StudentOptional = {"name": "Ravi"}
print(student_with_age)
print(student_no_age)
```

To make the key optional **and** allow the value to be `None`, combine `NotRequired` with a union:

```python
class StudentOptionalUnion(TypedDict):
    age: NotRequired[int | None]

print({"age": 21})
print({"age": None})
print({})
```

### 5. TypedDict in LangGraph
`TypedDict` is extremely useful in LangGraph to define a state that is built up step-by-step.

```python
class AgentState(TypedDict):
    question: str
    context: NotRequired[str]
    answer: NotRequired[str]

# Initially:
state: AgentState = {
    "question": "What is RAG?"
}
print("Initial:", state)

# After retrieval:
state["context"] = "RAG retrieves relevant information..."
print("After Retrieval:", state)

# After generation:
state["answer"] = "RAG combines retrieval with generation."
print("After Generation:", state)
```
