### Follow-Along Practice: Chat Model Basic
#### GoalCreate a model instance and query it for the result of a simple math calculation.

#### Method 1: Using Core Abstraction Helper (Recommended)1. Import `init_chat_model` from `langchain.chat_models`.
2. Initialize model using `init_chat_model("gpt-4o", model_provider="openai")`.
3. Call `model.invoke()` with the input string and print the response.

````python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv()

# Initialize using the core abstraction helper
model = init_chat_model("gpt-4o", model_provider="openai")

# Invoke the model
result = model.invoke("What is 81 divided by 9?")
print("Full result:")
print(result)
print("Content only:")
print(result.content)
````

#### Method 2: Using Proprietary Wrapper Class1. Import the proprietary model wrapper `ChatOpenAI` from `langchain_openai`.
2. Initialize the model using `ChatOpenAI(model="gpt-4o")`.
3. Call `model.invoke()` with the input string and print the response.

````python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI

load_dotenv()

# Initialize using the proprietary class
model = ChatOpenAI(model="gpt-4o")

# Invoke the model
result = model.invoke("What is 81 divided by 9?")
print("Full result:")
print(result)
print("Content only:")
print(result.content)
````
