### Agents & Tools - Practice & Exercises
In this notebook, you will practice constructing custom tools using LangChain's `@tool` decorator and letting a ReAct agent execute them.

#### Part 1: Follow-Along Practice
##### GoalCreate a custom string-length calculation tool and run a ReAct agent to inspect and resolve user queries.

##### Steps:1. Define the custom tool `@tool` with typing and docstring explanation.
2. Bind the tool to ChatOpenAI.
3. Setup prebuilt `create_react_agent` from `langgraph.prebuilt` (or a manual ReAct loop in LangChain).
4. Invoke the agent.

````python
from langchain_core.tools import tool
from langchain_openai import ChatOpenAI
from langgraph.prebuilt import create_react_agent
from dotenv import load_dotenv

load_dotenv()

@tool
def get_word_length(word: str) -> int:
    """Returns the number of letters in the provided word."""
    return len(word)

tools = [get_word_length]
model = ChatOpenAI(model="gpt-4o")
agent = create_react_agent(model, tools)

response = agent.invoke({"messages": [("user", "What is the length of the word 'Antigravity'?")]})
response["messages"][-1].pretty_print()
````

from langchain_core.tools import tool
from langchain_openai import ChatOpenAI
from langgraph.prebuilt import create_react_agent

@tool
def get_weather(city: str) -> str:
    """Returns the current weather temperature and conditions for a given city."""
    return f"The weather in {city} is 32C and Sunny today."

tools = [get_weather]
model = ChatOpenAI(model="gpt-4o")
agent = create_react_agent(model, tools)

res = agent.invoke({"messages": [("user", "What is the weather in Delhi today?")]})
res["messages"][-1].pretty_print()
