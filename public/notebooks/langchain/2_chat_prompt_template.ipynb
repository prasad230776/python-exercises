### ChatPromptTemplate - Practice & Exercises

In this notebook, you will practice formatting dynamic structured message prompts using LangChain's `ChatPromptTemplate` and invoking them using core abstractions (Groq provider).

#### Part 1: Follow-Along Practice

##### Method 1: Using `.from_messages()`
Define system behavior and human instructions using role/template tuples.

````python
# Load environment variables from .env file
from dotenv import load_dotenv
load_dotenv()

# Import ChatPromptTemplate from langchain
from langchain.prompts import ChatPromptTemplate
# Import core abstraction to initialize chat model
from langchain.chat_models import init_chat_model

# Initialize Groq model using core abstractions
model = init_chat_model("llama-3.3-70b-versatile", model_provider="groq")

# Define system and human dynamic templates using tuples
messages_tuples = [
    ("system", "You are a comedian who tells jokes about {topic}."),
    ("human", "Tell me {joke_count} jokes."),
]
# Create ChatPromptTemplate
chat_prompt_template = ChatPromptTemplate.from_messages(messages_tuples)

# Format and invoke the prompt template
prompt_value = chat_prompt_template.invoke({"topic": "lawyers", "joke_count": 3})
# Run the model with the formatted prompt value
response = model.invoke(prompt_value)
print(response.content)
````

##### Method 2: Direct Class Constructor
Initialize `ChatPromptTemplate` directly by passing list of message objects.

````python
# Import ChatPromptTemplate and message classes
from langchain.prompts import ChatPromptTemplate
from langchain_core.messages import SystemMessage, HumanMessage

# Create ChatPromptTemplate directly passing message structures
chat_prompt_direct = ChatPromptTemplate(
    messages=[
        SystemMessage(content="You are a helpful, polite assistant."),
        HumanMessage(content="Tell me what Python is.")
    ]
)

# Format and invoke directly
prompt_value = chat_prompt_direct.invoke({})
# Run the model
response = model.invoke(prompt_value)
print(response.content)
````

from langchain.prompts import ChatPromptTemplate
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv

load_dotenv()
model = init_chat_model("llama-3.3-70b-versatile", model_provider="groq")

# Define the ChatPromptTemplate using tuples
email_template = ChatPromptTemplate.from_messages([
    ("system", "You are a professional business writing assistant."),
    ("human", "Write a brief, polite email to {recipient_name} at {company} regarding: {reason}.")
])

# Invoke and run the template
prompt = email_template.invoke({
    "recipient_name": "Mr. Sunil",
    "company": "Tech Corp",
    "reason": "proposing a software demo"
})
response = model.invoke(prompt)
print(response.content)
