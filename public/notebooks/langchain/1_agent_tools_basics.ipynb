#### Part 1: Follow-Along Practice

##### Goal / Question
Create a ReAct agent with access to a local time tool, query it for the current time, and print the response.

##### Sample Input
```python
{"input": "What time is it?"}
```

##### Sample Output
A dictionary containing the agent response.

##### Plan / Steps
1. Define a helper function `get_current_time` returning time formatted as `H:MM AM/PM`.
2. Expose the function as a tool using `Tool(name="Time", func=get_current_time, description="...")`.
3. Pull prompt `hwchase17/react` from the hub.
4. Initialize model using `init_chat_model("llama-3.3-70b-versatile", model_provider="groq")`.
5. Create the agent with `create_react_agent` and execute it using `AgentExecutor.from_agent_and_tools`.

````python
from dotenv import load_dotenv
from langchain import hub
from langchain.agents import AgentExecutor, create_react_agent
from langchain_core.tools import Tool
from langchain.chat_models import init_chat_model
import datetime

load_dotenv()

def get_current_time(*args, **kwargs):
    now = datetime.datetime.now()
    return now.strftime("%I:%M %p")

tools = [
    Tool(
        name="Time",
        func=get_current_time,
        description="Useful for when you need to know the current time",
    )
]

prompt = hub.pull("hwchase17/react")
llm = init_chat_model("llama-3.3-70b-versatile", model_provider="groq", temperature=0)
agent = create_react_agent(llm=llm, tools=tools, prompt=prompt, stop_sequence=True)
agent_executor = AgentExecutor.from_agent_and_tools(agent=agent, tools=tools, verbose=True)
response = agent_executor.invoke({"input": "What time is it?"})
print(response)
````

def get_string_length(text: str, *args, **kwargs):
    return str(len(str(text).strip()))

length_tool = Tool(
    name="StringLength",
    func=get_string_length,
    description="Useful for calculating the character length of a string."
)
agent = create_react_agent(llm=llm, tools=[length_tool], prompt=prompt, stop_sequence=True)
executor = AgentExecutor.from_agent_and_tools(agent=agent, tools=[length_tool], verbose=True)
print(executor.invoke({"input": "Calculate the length of the string: LangChain"}))
