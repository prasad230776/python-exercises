### Follow-Along Practice: User Conversation Loop
#### GoalCreate a continuous conversation loop where the model queries input from the user and maintains chat memory in a local list.

#### Plan1. Initialize `init_chat_model` with a standard model.
2. Create a list `chat_history` with an initial `SystemMessage`.
3. Use a `while True` loop to fetch input, append `HumanMessage`, invoke the model, print the response, and append `AIMessage`.

````python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage

load_dotenv()

model = init_chat_model("gpt-4o", model_provider="openai")
chat_history = [SystemMessage(content="You are a helpful AI assistant.")]

print("Start chatting with the AI. Type 'exit' to quit.")
# Simulation of a loop (simulated inputs since we are in a notebook)
queries = [
    "Hello, I live in Seattle.",
    "What is my favorite sport? Oh, wait, I didn't tell you. My favorite sport is Soccer.",
    "Where do I live?",
    "What is my favorite sport?"
]

for query in queries:
    print(f"\nUser: {query}")
    chat_history.append(HumanMessage(content=query))
    result = model.invoke(chat_history)
    response = result.content
    chat_history.append(AIMessage(content=response))
    print(f"AI: {response}")
````

from langchain.chat_models import init_chat_model
