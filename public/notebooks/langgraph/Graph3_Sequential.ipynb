#### Part 1: Follow-Along Practice
##### GoalWe want to build a sequential graph with two nodes that runs in a fixed order:
```
START ──> first_node ──> second_node ──> END
```

##### Steps:1. Define the `AgentState` schema with `name`, `age`, and `final` variables.
2. Define `first_node` (to set greeting) and `second_node` (to append age description).
3. Build and compile the graph.
4. Invoke the graph and view the final result.

````python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END

# 1. Define the State Schema
class AgentState(TypedDict):
    name: str
    age: str
    final: str
````

````python
# 2. Define the Nodes
def first_node(state: AgentState) -> AgentState:
    state["final"] = f"Hi {state['name']}!"
    return state

def second_node(state: AgentState) -> AgentState:
    state["final"] = state["final"] + f" You are {state['age']} years old!"
    return state
````

````python
# 3. Create and Compile the Graph
graph = StateGraph(AgentState)

# Add both nodes
graph.add_node("first_node", first_node)
graph.add_node("second_node", second_node)

# Define the flow
graph.set_entry_point("first_node")
graph.add_edge("first_node", "second_node")
graph.set_finish_point("second_node")

app = graph.compile()
````

````python
# 4. Invoke the Graph
answers = app.invoke({"name": "Chirag", "age": 20})
print(answers["final"])
````
