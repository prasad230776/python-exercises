### 1. Traditional vs. GenAI Applications
Traditional software uses strict hardcoded logic and structured database tables. GenAI software uses unstructured natural language and probabilistic models.

### 2. Two Kinds of GenAI Applications
* **Sequential Workflows**: Step-by-step linear pipelines.
* **Agentic Workflows**: Autonomous loops where the LLM dynamically chooses tools at runtime.

#### 4.4 Unified LangChain Invocation (Active)

````python
from langchain.chat_models import init_chat_model

# Unified model connection wrapper
llm = init_chat_model("gemini-2.5-flash", model_provider="google_genai")
response = llm.invoke("Explain LangChain in 1 sentence.")
print(response.content)
````

### 7. Initializing and Calling Models

#### 7.1 Initializing with Groq

````python
from langchain.chat_models import init_chat_model

# Initialize the Groq model
groq_llm = init_chat_model(
    "llama-3.3-70b-versatile",
    model_provider="groq"
)

response = groq_llm.invoke("Explain why developers use virtual environments in Python.")
print(response.content)
````

#### 7.2 Initializing with Google Gemini

````python
from langchain.chat_models import init_chat_model

# Initialize the Gemini model
gemini_llm = init_chat_model(
    "gemini-2.5-flash",
    model_provider="google_genai"
)

response = gemini_llm.invoke("What is the difference between concurrency and parallelism?")
print(response.content)
````

Write a script that prompts **both** Groq (`llama-3.3-70b-versatile`) and Google (`gemini-2.5-flash`) with the question `"State the main goal of prompt engineering in 5 words."` and prints the response from each model.
