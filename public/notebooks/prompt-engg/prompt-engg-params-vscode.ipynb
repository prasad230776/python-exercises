### 1. Key Hyperparameters

* **Temperature**: Controls token selection randomness.
* **Max Tokens**: Bounds target length constraints.
* **Top-K**: Caps choices to the top K tokens.
* **Top-P**: Caps choices to cumulative probability P.

#### A. Low Temperature (Deterministic) Code Example

````python
from langchain.chat_models import init_chat_model

# Initialize with temperature=0.0
llm_strict = init_chat_model("gemini-2.5-flash", model_provider="google_genai", temperature=0.0)
response = llm_strict.invoke("Explain Python in 1 word.")
print("Deterministic output:", response.content)
````

#### B. High Temperature (Creative) Code Example

````python
# Initialize with temperature=1.2
llm_creative = init_chat_model("gemini-2.5-flash", model_provider="google_genai", temperature=1.2)
response = llm_creative.invoke("Create a wacky new name for a cat breed.")
print("Creative output:", response.content)
````

#### C. Max Tokens Constraint Code Example

````python
# Force the model to terminate generation early
response = llm_strict.invoke("Explain the concept of quantum computing in depth.", max_tokens=15)
print("Max token limited output:", response.content)
````
